```python
import cv2
import matplotlib.pyplot as plt
from google.colab import files

# Step 1: Upload image
uploaded = files.upload()
filename = list(uploaded.keys())[0]

# Step 2: Read image in grayscale
img_gray = cv2.imread(filename, cv2.IMREAD_GRAYSCALE)

# Step 3: Apply different thresholding methods

# 1️⃣ Simple Binary Threshold
_, thresh_binary = cv2.threshold(img_gray, 127, 255, cv2.THRESH_BINARY)

# 2️⃣ Binary Inverse
_, thresh_binary_inv = cv2.threshold(img_gray, 127, 255, cv2.THRESH_BINARY_INV)

# 3️⃣ Truncate
_, thresh_trunc = cv2.threshold(img_gray, 127, 255, cv2.THRESH_TRUNC)

# 4️⃣ ToZero
_, thresh_tozero = cv2.threshold(img_gray, 127, 255, cv2.THRESH_TOZERO)

# Step 4: Display the results
fig, axes = plt.subplots(1,5, figsize=(20,5))

axes[0].imshow(img_gray, cmap='gray')
axes[0].set_title("Original Grayscale")
axes[0].axis('off')

axes[1].imshow(thresh_binary, cmap='gray')
axes[1].set_title("Binary Threshold")
axes[1].axis('off')

axes[2].imshow(thresh_binary_inv, cmap='gray')
axes[2].set_title("Binary Inverse")
axes[2].axis('off')

axes[3].imshow(thresh_trunc, cmap='gray')
axes[3].set_title("Truncate")
axes[3].axis('off')

axes[4].imshow(thresh_tozero, cmap='gray')
axes[4].set_title("ToZero")
axes[4].axis('off')

plt.tight_layout()
plt.show()

# Step 5: Analyze segmentation
print("Analysis:")
print("- Binary threshold: Pixels > 127 become white, others black")
print("- Binary inverse: Pixels > 127 become black, others white")
print("- Truncate: Pixels > 127 clipped to 127, others unchanged")
print("- ToZero: Pixels <=127 set to 0, others unchanged")
```
