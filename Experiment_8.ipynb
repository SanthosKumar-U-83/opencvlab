```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
import zipfile
import os

# Step 1: Upload ZIP dataset
uploaded = files.upload()
zip_filename = list(uploaded.keys())[0]

# Step 2: Extract ZIP
dataset_dir = "dataset"
with zipfile.ZipFile(zip_filename, 'r') as zip_ref:
    zip_ref.extractall(dataset_dir)

# Step 3: List all image files
image_files = []
for root, dirs, files_list in os.walk(dataset_dir):
    for file in files_list:
        if file.lower().endswith(('.png', '.jpg', '.jpeg', '.bmp')):
            image_files.append(os.path.join(root, file))

print(f"Found {len(image_files)} images in dataset.")

# Define a 3x3 kernel for morphological operations
kernel = np.ones((3,3), np.uint8)

# Step 4: Process each image
for img_path in image_files:
    # Load image
    img = cv2.imread(img_path)
    img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # Threshold to binary
    _, img_bin = cv2.threshold(img_gray, 127, 255, cv2.THRESH_BINARY)

    # Morphological operations
    erosion = cv2.erode(img_bin, kernel, iterations=1)
    dilation = cv2.dilate(img_bin, kernel, iterations=1)
    opening = cv2.morphologyEx(img_bin, cv2.MORPH_OPEN, kernel)
    closing = cv2.morphologyEx(img_bin, cv2.MORPH_CLOSE, kernel)
    gradient = cv2.morphologyEx(img_bin, cv2.MORPH_GRADIENT, kernel)

    # Display results in a 2x3 grid
    fig, axes = plt.subplots(2,3, figsize=(15,10))

    axes[0,0].imshow(img_gray, cmap='gray')
    axes[0,0].set_title("Grayscale")
    axes[0,0].axis('off')

    axes[0,1].imshow(img_bin, cmap='gray')
    axes[0,1].set_title("Binary")
    axes[0,1].axis('off')

    axes[0,2].imshow(erosion, cmap='gray')
    axes[0,2].set_title("Erosion")
    axes[0,2].axis('off')

    axes[1,0].imshow(dilation, cmap='gray')
    axes[1,0].set_title("Dilation")
    axes[1,0].axis('off')

    axes[1,1].imshow(opening, cmap='gray')
    axes[1,1].set_title("Opening")
    axes[1,1].axis('off')

    axes[1,2].imshow(closing, cmap='gray')
    axes[1,2].set_title("Closing")
    axes[1,2].axis('off')

    plt.suptitle(f"Morphological Operations: {os.path.basename(img_path)}", fontsize=16)
    plt.tight_layout()
    plt.show()

    # Optional: save processed images
    # out_dir = "processed_images"
    # os.makedirs(out_dir, exist_ok=True)
    # cv2.imwrite(os.path.join(out_dir, f"erosion_{os.path.basename(img_path)}"), erosion)
    # cv2.imwrite(os.path.join(out_dir, f"dilation_{os.path.basename(img_path)}"), dilation)
    # ... similarly for opening, closing, gradient
```
