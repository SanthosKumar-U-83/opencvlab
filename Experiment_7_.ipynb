```python
import cv2
import matplotlib.pyplot as plt
from google.colab import files

# Step 1: Upload image
uploaded = files.upload()
filename = list(uploaded.keys())[0]

# Step 2: Read image in grayscale
img_gray = cv2.imread(filename, cv2.IMREAD_GRAYSCALE)

# Step 3: Apply Sobel Edge Detection
sobel_x = cv2.Sobel(img_gray, cv2.CV_64F, 1, 0, ksize=3)  # X direction
sobel_y = cv2.Sobel(img_gray, cv2.CV_64F, 0, 1, ksize=3)  # Y direction
sobel_combined = cv2.magnitude(sobel_x, sobel_y)  # Combine X and Y

# Convert back to uint8
sobel_combined = cv2.convertScaleAbs(sobel_combined)

# Step 4: Apply Laplacian Edge Detection
laplacian = cv2.Laplacian(img_gray, cv2.CV_64F)
laplacian = cv2.convertScaleAbs(laplacian)

# Step 5: Apply Canny Edge Detection
canny = cv2.Canny(img_gray, 100, 200)  # threshold1=100, threshold2=200

# Step 6: Display all outputs
fig, axes = plt.subplots(1,4, figsize=(20,5))

axes[0].imshow(img_gray, cmap='gray')
axes[0].set_title("Original Grayscale")
axes[0].axis('off')

axes[1].imshow(sobel_combined, cmap='gray')
axes[1].set_title("Sobel Edge")
axes[1].axis('off')

axes[2].imshow(laplacian, cmap='gray')
axes[2].set_title("Laplacian Edge")
axes[2].axis('off')

axes[3].imshow(canny, cmap='gray')
axes[3].set_title("Canny Edge")
axes[3].axis('off')

plt.tight_layout()
plt.show()
```
